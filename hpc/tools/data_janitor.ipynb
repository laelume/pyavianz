## Schlepping data from one place to another

```python
file = r'D:\acoustic_01_files.csv'
```

```python
import pandas as pd

df = pd.read_csv(r"D:\acoustic_01_files.csv")

alberto = df[df["path"].str.contains("/AlbertoData/", regex=False)]

print(f"{len(alberto)} of {len(df)} files are under AlbertoData")
print(alberto.head())

alberto.to_csv(r"D:\acoustic_01_AlbertoData.csv", index=False)
```

```python
filename = alberto["path"].str.rsplit("/", n=1).str[-1]
hour = filename.str.extract(r"\d{6,8}_(\d{2})\d{4}", expand=False).astype(float)

night = alberto[(hour >= 19) | (hour < 7)]

print(f"{len(night)} of {len(alberto)} files start between 19:00 and 06:59")
print(f"{hour.isna().sum()} files have no datetime in the name (excluded)")
print(night.head())

night.to_csv(r"D:\acoustic_01_AlbertoData_night.csv", index=False)
```
